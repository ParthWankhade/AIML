```python
import random


class Environment:
    def __init__(self):
        self.rooms = {
            "Floor 1": {
                "A": random.randint(0, 1),
                "B": random.randint(0, 1),
                "C": random.randint(0, 1)
            },
            "Floor 2": {
                "A": random.randint(0, 1),
                "B": random.randint(0, 1),
                "C": random.randint(0, 1)
            },
            "Floor 3": {
                "A": random.randint(0, 1),
                "B": random.randint(0, 1),
                "C": random.randint(0, 1)
            }
        }


class VacuumAgent:
    def __init__(self, environment):

        self.environment = environment

        self.score = 0
        self.battery = 100

        self.max_battery = 100
        self.cleaning_cost = 20
        self.movement_cost = 10

        # Charging station is Room A on every floor
        self.charging_rooms = {
            "Floor 1": "A",
            "Floor 2": "A",
            "Floor 3": "A"
        }

        # Starting position
        self.current_floor = "Floor 1"
        self.current_room = "A"

        print("=" * 55)
        print("INITIAL ENVIRONMENT")
        print("=" * 55)

        self.display_environment()

        print("\nCharging Stations:")
        print("Floor 1 - Room A")
        print("Floor 2 - Room A")
        print("Floor 3 - Room A")

        print("\nInitial Battery:", self.battery, "%")
        print("Starting Position: Floor 1 - Room A")

        self.clean_environment()

        print("\n" + "=" * 55)
        print("FINAL RESULT")
        print("=" * 55)

        self.display_environment()

        print("\nFinal Position:",
              self.current_floor,
              "Room",
              self.current_room)

        print("Final Battery:", self.battery, "%")
        print("Performance Score:", self.score)


    # Display all rooms
    def display_environment(self):

        for floor, rooms in self.environment.rooms.items():

            print("\n" + floor + ":")

            for room, condition in rooms.items():

                if condition == 1:
                    status = "Dirty"
                else:
                    status = "Clean"

                print("Room", room, ":", status)


    # Convert floor name into number
    def get_floor_number(self, floor):

        if floor == "Floor 1":
            return 1
        elif floor == "Floor 2":
            return 2
        else:
            return 3


    # Convert room name into number
    def get_room_number(self, room):

        if room == "A":
            return 0
        elif room == "B":
            return 1
        else:
            return 2


    # Calculate number of movement steps
    def calculate_movement_steps(self,
                                 from_floor,
                                 from_room,
                                 to_floor,
                                 to_room):

        floor_difference = abs(
            self.get_floor_number(from_floor)
            -
            self.get_floor_number(to_floor)
        )

        room_difference = abs(
            self.get_room_number(from_room)
            -
            self.get_room_number(to_room)
        )

        return floor_difference + room_difference


    # Move vacuum to target location
    def move_to(self, target_floor, target_room):

        print("\nMoving from",
              self.current_floor,
              "Room",
              self.current_room,
              "to",
              target_floor,
              "Room",
              target_room)

        # Move between floors
        current_floor_number = self.get_floor_number(
            self.current_floor
        )

        target_floor_number = self.get_floor_number(
            target_floor
        )

        while current_floor_number != target_floor_number:

            if target_floor_number > current_floor_number:
                current_floor_number += 1
            else:
                current_floor_number -= 1

            self.current_floor = "Floor " + str(current_floor_number)

            self.battery -= self.movement_cost

            print("Moved to", self.current_floor,
                  "| Battery:", self.battery, "%")


        # Move between rooms
        current_room_number = self.get_room_number(
            self.current_room
        )

        target_room_number = self.get_room_number(
            target_room
        )

        while current_room_number != target_room_number:

            if target_room_number > current_room_number:
                current_room_number += 1
            else:
                current_room_number -= 1

            if current_room_number == 0:
                self.current_room = "A"
            elif current_room_number == 1:
                self.current_room = "B"
            else:
                self.current_room = "C"

            self.battery -= self.movement_cost

            print("Moved to",
                  self.current_floor,
                  "Room",
                  self.current_room,
                  "| Battery:",
                  self.battery,
                  "%")


    # Charge battery on the current floor
    def go_to_charging_station(self):

        previous_floor = self.current_floor
        previous_room = self.current_room

        # Charging station of current floor
        charging_room = self.charging_rooms[
            self.current_floor
        ]

        print("\n*** BATTERY LOW ***")
        print("Current Battery:", self.battery, "%")

        print("Charging station on",
              self.current_floor,
              "is Room",
              charging_room)

        # Move to charging station
        self.move_to(
            self.current_floor,
            charging_room
        )

        # Charge battery
        print("\nCharging...")
        self.battery = self.max_battery
        print("Battery charged to", self.battery, "%")

        # Return to previous location
        if previous_room != charging_room:

            print("\nReturning to previous location...")

            self.move_to(
                previous_floor,
                previous_room
            )

        print("Returned to",
              self.current_floor,
              "Room",
              self.current_room)


    # Clean one room
    def clean_room(self, floor, room):

        condition = self.environment.rooms[floor][room]

        if condition == 1:

            print("Room", room, "is Dirty")
            print("Cleaning Room", room)

            self.environment.rooms[floor][room] = 0

            self.battery -= self.cleaning_cost

            self.score += 1

            print("Room", room, "cleaned")
            print("Battery:", self.battery, "%")
            print("Score:", self.score)

        else:

            print("Room", room,
                  "is already Clean")
            print("No cleaning required")


    # Main cleaning process
    def clean_environment(self):

        for floor_number in range(1, 4):

            floor_name = "Floor " + str(floor_number)

            for room in ["A", "B", "C"]:

                # Calculate movement required
                required_steps = self.calculate_movement_steps(
                    self.current_floor,
                    self.current_room,
                    floor_name,
                    room
                )

                required_battery = (
                    required_steps *
                    self.movement_cost
                )

                print("\n---------------------------------------")
                print("Target:", floor_name,
                      "Room", room)

                print("Current Battery:",
                      self.battery, "%")

                print("Required Movement Battery:",
                      required_battery, "%")

                # Check whether battery is enough
                if self.battery < required_battery:

                    self.go_to_charging_station()

                # Move to target room
                if not (
                    self.current_floor == floor_name
                    and
                    self.current_room == room
                ):

                    self.move_to(
                        floor_name,
                        room
                    )

                # Clean room
                self.clean_room(
                    floor_name,
                    room
                )


# --------------------------------------------------
# MAIN PROGRAM
# --------------------------------------------------

environment = Environment()

vacuum = VacuumAgent(environment)
```

**RESULT ANALYSIS**: The vacuum cleaner successfully operated in an environment consisting of 3 floors and 3 rooms on each floor, making a total of 9 rooms.     
Each room was randomly assigned a Clean or Dirty condition, allowing the agent to make decisions based on the room's current state.                             
The agent successfully identified and cleaned the dirty rooms, while clean rooms were skipped.                                                              
Battery consumption was successfully managed, with 10% battery used for each movement step and 20% battery used for cleaning a dirty room.                    
When the battery level became insufficient, the vacuum automatically moved to Room A of the current floor, which acted as the charging station.                
After charging to 100%, the vacuum returned to its previous location and continued the cleaning process.                                                  
The performance score increased by 1 for every successfully cleaned room, providing a measure of the agent's performance.                                  
The final output displayed the room conditions, final position, remaining battery, and performance score.                                                  
Thus, the experiment demonstrated that the vacuum agent can perform autonomous cleaning along with battery management and floor-based charging.
